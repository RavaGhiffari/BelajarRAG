```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

model = SentenceTransformer("BAAI/bge-small-en-v1.5")
```

```python
documents = [
    "Spiderman is Peter Parker.",
    "Ironman made the Avengers Tower.",
    "Ironman owns Stark Industries.",
    "Thor is the God of Thunder."
]

document_embeddings = ask_documents(documents)
```

```python
def ask_documents(documents):
    return model.encode(documents)

def inference(question, documents, document_embeddings):
    query_embedding = model.encode(question)

    scores = cosine_similarity(
        [query_embedding],
        document_embeddings
    )[0]

    index = np.argmax(scores)

    # return {
    #     "question": question,
    #     "answer": documents[index],
    #     "score": scores[index]
    # }

    return scores

```

```python
scores = inference(
    "Who made the Avengers tower?",
    documents,
    document_embeddings
)

for doc, score in zip(documents, scores):
    print(f"{score:.4f} -> {doc}")
```
